```python
import pandas as pd
from pathlib import Path
```

```python
file_path = Path("..") / ".." / "datalake" / "raw_data" / "stock_prices_daily_from_2010.csv"
file_path = file_path.resolve()
```

```python
df = pd.read_csv(file_path, header=[0, 1, 2])
```

```python
kolumna_z_data = ('Price', 'Ticker', 'Date')
if kolumna_z_data in df.columns:
    df.set_index(kolumna_z_data, inplace=True)
    df.index.name = 'Date'  # Uporządkowanie nazwy indeksu

czy_sa_braki = df.isnull().values.any()
print(f"Czy w danych są braki? {czy_sa_braki}\n")

if czy_sa_braki:
    print("Zestawienie brakujących dat dla poszczególnych spółek i wskaźników:\n" + "-"*60)
    
    # Przeszukujemy każdą kolumnę pod kątem wartości NaN
    for col in df.columns:
        missing_mask = df[col].isna()
        
        if missing_mask.any():
            # Ekstrakcja dat (indeksów), dla których brakuje danych
            missing_dates = df.index[missing_mask].tolist()
            
            # Pobranie metryki (np. Close, High) i nazwy spółki (np. META, AAPL)
            metric = col[0]
            ticker = col[1]
            
            print(f"Spółka: {ticker} | Wskaźnik: {metric}")
            print(f"Liczba brakujących dni: {len(missing_dates)}")
            
            # Formatowanie wyświetlania dat
            if len(missing_dates) > 10:
                print(f"Daty: {missing_dates[:3]} ... [pominięto {len(missing_dates)-6}] ... {missing_dates[-3:]}\n")
            else:
                print(f"Daty: {missing_dates}\n")
else:
    print("Zbiór danych jest kompletny.")
```
